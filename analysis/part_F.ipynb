# Part F — Budget Reallocation Plan & 30-Day Impact Estimate

Budget Assumption

Total Monthly Marketing Budget: ₹10,00,000

## Proposed Budget Allocation

| Channel       | Current % | New % | Change |
|--------------|----------|------|--------|
| Search       | 46%      | 40%  | ↓ Slight decrease |
| Paid Social  | 38%      | 20%  | 🔻 Significant reduction |
| Email        | 5%       | 15%  | 🔺 Increase |
| Organic      | 1%       | 10%  | 🔺 Increase |
| Referral     | 9%       | 15%  | 🔺 Increase |

## Rationale

Budget is reallocated from Paid Social to Email, Organic, and Referral channels based on insights from attribution and regression analysis.

Email shows strong efficiency (high ROAS)

Organic delivers high revenue with minimal cost

Referral shows the highest marginal impact in regression

Search remains a key revenue driver and is retained with slight optimization

Paid Social shows low efficiency and is reduced but not eliminated

## Constraints
Minimum 10% spend maintained for key acquisition channels (e.g., Search)

Maximum 40% cap to avoid over-dependence on a single channel

Paid Social is retained at a lower level to allow for testing and optimization

## Expected Impact (30-Day Estimate)
Base Case: Revenue expected to increase by 8–12%

Best Case: Revenue increase of 15–20% if efficiencies hold

Worst Case: Revenue increase of 3–5% due to diminishing returns or external factors

## Assumptions
Historical performance trends remain stable

Channel efficiencies (ROAS and marginal effects) continue in the short term

No major external disruptions (pricing, competition, demand shifts)

## Sensitivity

The results are sensitive to model assumptions, especially regression coefficients and attribution logic. Changes in campaign quality, seasonality, or customer behavior may impact actual outcomes.

## Final Recommendation

Budget should be shifted away from Paid Social toward Email, Referral, and Organic channels, while maintaining Search as a core revenue driver. This reallocation is expected to improve overall marketing efficiency and increase revenue in the next month.
